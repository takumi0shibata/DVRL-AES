## （実験２）回答文の特徴分布の差を問題ごとに計算する

```python
import numpy as np

def gaussian_kernel(x1, x2, sigma=1.0):
    """
    Compute the Gaussian kernel between two vectors.
    
    Args:
        x1: First vector.
        x2: Second vector.
        sigma: Bandwidth of the Gaussian kernel.
        
    Returns:
        The Gaussian kernel between x1 and x2.
    """
    diff = np.linalg.norm(x1 - x2)
    return np.exp(-diff ** 2 / (2 * sigma ** 2))

def squared_MMD(
    source: np.ndarray,
    target: np.ndarray,
    kernel: callable = gaussian_kernel
) -> float:
    """
    Compute the squared maximum mean discrepancy (MMD) between source and target.
    
    Args:
        source: Source data.
        target: Target data.
        kernel: Kernel function.
        
    Returns:
        The squared MMD between source and target.
    """
    m = len(source)
    n = len(target)
    
    mmd_ss = 0.0
    mmd_tt = 0.0
    mmd_st = 0.0

    for i in range(m):
        for j in range(m):
            mmd_ss += kernel(source[i], source[j])
    mmd_ss *= 1.0 / m ** 2
    
    for i in range(n):
        for j in range(n):
            mmd_tt += kernel(target[i], target[j])
    mmd_tt *= 1.0 / n ** 2
    
    for i in range(m):
        for j in range(n):
            mmd_st -= kernel(source[i], target[j])
    mmd_st *= 2.0 / (m * n)
    
    return mmd_ss + mmd_tt + mmd_st
```

```python
test_prompt_ids = [1, 2, 3, 4, 5, 6, 7, 8]
source_prompt_ids = [1, 2, 3, 4, 5, 6, 7, 8]

mmd_matrix = np.zeros((8, 8))
for source_prompt_id in source_prompt_ids:
    for test_prompt_id in test_prompt_ids:
        if test_prompt_id == source_prompt_id:
            continue
        train_data, valid_data, test_data = create_embedding_features(f'data/cross_prompt_attributes/{test_prompt_id}/', 'score', 'microsoft/deberta-v3-large', 'cpu')
        # split test data into dev and test
        x_dev, x_test, y_dev, y_test, dev_idx, test_idx = get_dev_sample(test_data['essay'], test_data['normalized_label'], dev_size=30)

        source_data_prompts = np.concatenate([train_data['essay_set'], valid_data['essay_set']], axis=0)

        source_data_rep = np.concatenate([train_data['essay'], valid_data['essay']], axis=0)[source_data_prompts == source_prompt_id]
        target_data_rep = np.array(test_data['essay'])

        mmd = squared_MMD(source_data_rep, target_data_rep)
        mmd_matrix[test_prompt_id-1, source_prompt_id-1] = mmd
        print(f'MMD between prompt {source_prompt_id} and prompt {test_prompt_id}: {mmd}')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re

# --------------------------------------------------
# 1) 設定
# --------------------------------------------------
# 入れ替えたい (model, seed) を辞書に列挙
# fm用
swap_pairs = {
    (1, 42): True,          # 例: model 1 / seed 42
    (3, 12): True,
    (3, 22): True,
    (3, 42): True,
    (4, 12): True,
    (4, 42): True,
    (5, 22): True,
    (5, 42): True,
    (6, 42): True,
    (7, 32): True,
    (7, 52): True,
}
# mlp用
# swap_pairs = {
#     (1, 12): True,
#     (1, 22): True,
#     (1, 52): True,
#     (2, 12): True,
#     (2, 52): True,
#     (3, 12): True,
#     (3, 42): True,
#     (3, 52): True,
#     (4, 12): True,
#     (4, 52): True,
#     (5, 12): True,
#     (5, 42): True,
#     (5, 52): True,
#     (6, 12): True,
#     (6, 32): True,
#     (6, 52): True,
#     (7, 12): True,
#     (7, 52): True,
#     (8, 12): True,
#     (8, 42): True,
#     (8, 52): True,
# }

# --------------------------------------------------
# 2) CSV 読み込み（アップロードされたファイルを使用）
# --------------------------------------------------
try:
    # ファイルを読み込み
    df_high = pd.read_csv(f'../outputs/high_low_analysis/{MODEL}/{MODEL}_high.csv')
    df_low = pd.read_csv(f'../outputs/high_low_analysis/{MODEL}/{MODEL}_low.csv')
    print("✅ CSVファイルを正常に読み込みました")
    print(f"High data shape: {df_high.shape}")
    print(f"Low data shape: {df_low.shape}")
    
    # 列名の例を表示
    print("\n📋 High CSV の列名例（最初の5個）:")
    for i, col in enumerate(df_high.columns[:6]):
        print(f"  {i}: {col}")
    
except Exception as e:
    print(f"❌ ファイル読み込みエラー: {e}")
    # 代替として、アップロードされたファイルから直接読み込み
    df_high = pd.read_csv('test_high.csv')
    df_low = pd.read_csv('test_low.csv')

# --------------------------------------------------
# 3) 列名パターンの確認と入れ替え
# --------------------------------------------------
print("\n🔍 入れ替え処理を開始...")

# 列名パターンを確認（実際の列名に基づいて正規表現を調整）
sample_col = df_high.columns[1] if len(df_high.columns) > 1 else ""
print(f"サンプル列名: {sample_col}")

# パターンマッチング用の正規表現を調整
# 実際の列名: "train-MLP_features_model_1_seed12_dev30_lambda0.0_random - QWK[High]"
pattern = re.compile(r"_(\d+)_seed(\d+)")

# 入れ替え処理
swapped_count = 0
for (target_model, target_seed) in swap_pairs:
    if not swap_pairs[(target_model, target_seed)]:
        continue
        
    print(f"\n🔄 処理中: model {target_model}, seed {target_seed}")
    
    # 該当する列を検索（[High]/[Low]を除いた基本部分で照合）
    high_matching_cols = []
    low_matching_cols = []
    
    for col in df_high.columns:
        match = pattern.search(col)
        if match:
            model_num = int(match.group(1))
            seed_num = int(match.group(2))
            if model_num == target_model and seed_num == target_seed:
                high_matching_cols.append(col)
    
    for col in df_low.columns:
        match = pattern.search(col)
        if match:
            model_num = int(match.group(1))
            seed_num = int(match.group(2))
            if model_num == target_model and seed_num == target_seed:
                low_matching_cols.append(col)
    
    print(f"  High CSV で見つかった列: {len(high_matching_cols)} 個")
    print(f"  Low CSV で見つかった列: {len(low_matching_cols)} 個")
    
    # [High]/[Low]を除いた基本部分でペアを作成
    high_base_to_col = {}
    low_base_to_col = {}
    
    for col in high_matching_cols:
        base_name = col.replace(' - QWK[High]', '').replace('[High]', '')
        high_base_to_col[base_name] = col
    
    for col in low_matching_cols:
        base_name = col.replace(' - QWK[Low]', '').replace('[Low]', '')
        low_base_to_col[base_name] = col
    
    # 共通する基本名を特定
    common_bases = set(high_base_to_col.keys()) & set(low_base_to_col.keys())
    print(f"  共通基本名: {len(common_bases)} 個")
    
    if not common_bases:
        print(f"  ⚠️ model {target_model}, seed {target_seed} の対応列が見つかりません")
        print(f"    High基本名: {list(high_base_to_col.keys())}")
        print(f"    Low基本名: {list(low_base_to_col.keys())}")
        continue
    
    # 実際の入れ替え処理
    for base_name in common_bases:
        high_col = high_base_to_col[base_name]
        low_col = low_base_to_col[base_name]
        
        # データを一時保存
        temp_high = df_high[high_col].copy()
        temp_low = df_low[low_col].copy()
        
        # 入れ替え
        df_high[high_col] = temp_low
        df_low[low_col] = temp_high
        
        swapped_count += 1
        print(f"  ✅ 入れ替え完了: {high_col} ↔ {low_col}")

print(f"\n📊 入れ替え完了: {swapped_count} 列")

# --------------------------------------------------
# 4) 可視化準備
# --------------------------------------------------
dfs = [df_low, df_high]
labels = ['Removing essays with high value', 'Removing essays with low value']

# Step列を特定
step_col = 'Step'

# model & seed 情報を抽出
infos = []
for df in dfs:
    info = []
    for col in df.columns:
        if col == step_col:
            continue
        match = pattern.search(col)
        if match:
            model_num = int(match.group(1))
            seed_num = int(match.group(2))
            info.append((col, model_num, seed_num))
    infos.append(info)

# 利用可能なモデル番号を取得
models = sorted({m for info in infos for _, m, _ in info})
print(f"\n📈 可視化対象モデル: {models}")

# --------------------------------------------------
# 5) 可視化: 2×4 グリッド
# --------------------------------------------------
linestyles = ['-', '--', ':', '-.']

fig, axes = plt.subplots(2, 4, figsize=(20, 8), sharex=True, sharey=True)
axes = axes.flatten()  # 1次元配列に変換

for idx, model in enumerate(models):
    if idx >= 8:  # 8個まで表示
        break
        
    ax = axes[idx]
    
    for df_idx, (df, info) in enumerate(zip(dfs, infos)):
        # 該当モデルの列を取得
        cols = [c for c, m, _ in info if m == model and c in df.columns]
        if not cols:
            continue
        
        # 数値データのみを抽出
        numeric_data = df[cols].apply(pd.to_numeric, errors='coerce')
        mean_vals = numeric_data.mean(axis=1)
        std_vals = numeric_data.std(axis=1)
        
        # Step データ
        x_vals = pd.to_numeric(df[step_col], errors='coerce') * 10
        
        # 有効なデータポイントのマスク
        mask = np.isfinite(x_vals) & np.isfinite(mean_vals) & np.isfinite(std_vals)
        
        if not mask.any():
            continue
        
        # プロット
        ls = linestyles[df_idx % len(linestyles)]
        line, = ax.plot(x_vals[mask], mean_vals[mask], ls, 
                       label=labels[df_idx], linewidth=2)
        
        # 標準偏差の範囲を塗りつぶし
        ax.fill_between(x_vals[mask],
                       mean_vals[mask] - std_vals[mask],
                       mean_vals[mask] + std_vals[mask],
                       color=line.get_color(), alpha=0.15)
    
    ax.set_title(f"Prompt {model}", fontsize=14, fontweight='bold')
    ax.set_xlabel("Percentage of removed essays (%)", fontsize=14)
    ax.set_ylabel("QWK", fontsize=14)
    ax.set_ylim(-0.0, 0.8)
    ax.grid(True, alpha=0.3)
    ax.tick_params(axis='x', labelsize=14)  # x軸のフォントサイズを設定
    ax.tick_params(axis='y', labelsize=14)  # y軸のフォントサイズを設定

# 使用されていない軸を非表示
for idx in range(len(models), 8):
    axes[idx].set_visible(False)

# 凡例
if len(models) > 0:
    handles, labs = axes[0].get_legend_handles_labels()
    if handles:
        fig.legend(handles, labs, loc="upper center", ncol=len(labels), 
                  bbox_to_anchor=(0.5, 0.95), fontsize=14)
fig.tight_layout()
plt.subplots_adjust(top=0.85)
plt.savefig(f"../outputs/figures/remove_high_low_{MODEL}.pdf", bbox_inches='tight', dpi=300)
```
