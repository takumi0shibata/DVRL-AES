```python
import json
import polars as pl
```

```python
results = {}
for d_size in [10, 20, 30, 40, 50, 100, 200, 500]:
    qwks = []
    for i in range(1, 9):
        with open(f'devonly_{i}_{d_size}.json', 'r') as f:
            data = json.load(f)
            qwk = data['eval_qwk']
        qwks.append(qwk)
    results[str(d_size)] = qwks
df = pl.DataFrame(results)
```
